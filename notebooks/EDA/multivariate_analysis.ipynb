```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
house_prices = pd.read_csv('../../datasets/house_prices_cleaned.zip')
house_copy = house_prices.copy()
house_copy.drop(columns= 'Unnamed: 0', axis= 'columns', inplace= True)
```

```python
house_copy.info()
```

```python
location_count = house_copy['location'].value_counts().head(22)
location_df = house_copy[house_copy['location'].isin(location_count.index)]
ax = sns.scatterplot(data= location_df, x= 'carpet_area', y= 'amount', hue= 'location', edgecolor= 'black')
plt.xscale('log')
plt.yscale('log')
ax.get_legend().remove()
```

```python
society_count = house_copy['society'].value_counts().head(24)
society_df = house_copy[house_copy['society'].isin(society_count.index)]
ax = sns.scatterplot(data= society_df, x= 'carpet_area', y= 'amount', hue= 'society', edgecolor= 'black')
plt.xscale('log')
plt.yscale('log')
ax.get_legend().remove()
```
