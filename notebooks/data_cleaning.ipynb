# Column Description

| **Column Name** | **Description**                                                                                           |
| --------------- | --------------------------------------------------------------------------------------------------------- |
| **Index**       | Unique serial number assigned to each property.                                                           |
| **Title**       | Short title or headline describing the property listing.                                                  |
| **Description** | Detailed textual description of the property and its features.                                            |
| **Amount**      | Total listed price of the property.                                                                       |
| **Price**       | Price per square foot of the property.                                                                    |
| **Location**    | City or locality where the property is located.                                                           |
| **Carpet Area** | Actual usable floor area inside the property, measured in square feet (sq ft) or square yards (sq yd).    |
| **Status**      | Indicates whether the property is ready for immediate occupancy or is under construction.                 |
| **Floor**       | Floor number on which the property is located.                                                            |
| **Transaction** | Indicates whether the property is a new property or a resale.                                             |
| **Furnishing**  | Furnishing status of the property (e.g., Furnished, Semi-Furnished, Unfurnished).                         |
| **Facing**      | Direction the property's main entrance or balcony faces (e.g., North, South, East, West).                 |
| **Overlooking** | Primary view from the property, such as a park, garden, pool, or main road.                               |
| **Society**     | Name of the residential society or apartment complex where the property is located.                       |
| **Bathroom**    | Total number of bathrooms in the property.                                                                |
| **Balcony**     | Total number of balconies available in the property.                                                      |
| **Car Parking** | Number of dedicated parking spaces available for vehicles.                                                |
| **Ownership**   | Type of legal ownership, such as Freehold or Cooperative Society.                                         |
| **Super Area**  | Total saleable area, including the carpet area, wall thickness, and a proportional share of common areas. |
| **Dimensions**  | Physical dimensions of the property (e.g., length × width).                                               |
| **Plot Area**   | Total land area occupied by the property.                                                                 |

# Data Cleaning

### Messy Data

- Extra index column is given.

- Two informations are in the same column Title which are number of bedrooms and location.

- The phrase `BHK Ready to Occupy Flat for Sale` is common in every value of the column Title.

- Description has 5 major informations number of bedrooms, precise area, price, commercials and whether it is furnished, semi-funished or unfurnished.

- Location has two informations total floors in buildings and on which floor flat is located.

- 63% of data is duplicated.

### Dirty Data

- Some column names are lower case and some are capitalize. `consisttency`

- Some of the values are same in Title and Description column. `uniqueness`

- Amount column has dtype string and inconsistent units. `inconsistency`

- Description, Price, Location, Status, facting, overlooking, society, balcony, car parking, ownership and has missing values. `completeness`

- Carpet Area and Super Area column has inconsitent units and missing values. `incosnistency` and `completeness`

- Dimensions and Plot Area has all values missing. `completeness`

- Bathroom and Balcony has dtype string instead of integer. `invalid`


#### Quality Dimensions Solution:
1. Uniqueness
2. Completeness
3. Tidiness
4. Validity
5. Accuracy 
6. Consistency
