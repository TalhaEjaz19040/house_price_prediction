# Data Cleaning

### Messy Data

- Extra index column is given.

- Two informations are in the same column Title which are number of bedrooms and location.

- The phrase `BHK Ready to Occupy Flat for Sale` is common in every value of the column Title.

- Description has 5 major informations number of bedrooms, precise area, price, commercials and whether it is furnished, semi-funished or unfurnished.

- Location has two informations total floors in buildings and on which floor flat is located.

- 63% of data is duplicated.

### Dirty Data

- Some column names are lower case and some are capitalize. `consisttency`

- Some of the values are same in Title and Description column. `uniqueness`

- Amount column has dtype string and inconsistent units. `inconsistency`

- Description, Price, Location, Status, facting, overlooking, society, balcony, car parking, ownership and has missing values. `completeness`

- Carpet Area and Super Area column has inconsitent units and missing values. `incosnistency` and `completeness`

- Dimensions and Plot Area has all values missing. `completeness`

- Bathroom and Balcony has dtype string instead of integer. `invalid`


#### Quality Dimensions Solution:
1. Uniqueness
2. Completeness
3. Tidiness
4. Validity
5. Accuracy 
6. Consistency

```python
house_copy['title'] = house_copy['title'].str.replace('BHK Ready to Occupy Flat for sale ', '')
house_copy['title'] = house_copy['title'].str.replace(' in', '')
house_copy['bedrooms'] = house_copy['title'].str.split(pat= ' ', n= 1).str.get(0)
house_copy['title'] = house_copy['title'].str.split(pat= ' ', n= 1).str.get(1)

def my_logic(s1, s2):
    for title, society in zip(s1, s2):
        if title is not np.nan and society is not np.nan:
            title = title.replace(society, '')
        yield title

func = my_logic(house_copy['title'], house_copy['society'])

house_copy['locality'] = pd.Series([new for new in func])

house_copy[(house_copy['locality'].isna() == False) & (house_copy['society'].isna() == False)].shape[0] / len(house_copy)
```

```python
house_copy['bedrooms'] = pd.to_numeric(house_copy['bedrooms'], errors= 'coerce')
```

```python
house_copy['amount'] = np.where((house_copy['amount'].str.contains('Call for Price')) | (house_copy['amount'].isna()), house_copy['description'].str.extract(r'(\d+\.?\d*\s*(?:Cr|Lac))', expand= False), house_copy['amount'])
```

```python
house_copy['figure'] = house_copy['amount'].str.split(' ').str.get(1).str.strip()
house_copy['amount'] = house_copy['amount'].str.split(' ').str.get(0).str.strip()
house_copy['amount'] = house_copy['amount'].astype('float64')
house_copy['amount'] = np.where(house_copy['figure'] == 'Cr', house_copy['amount'] * 100, house_copy['amount'])
print(house_copy[['figure', 'amount']])
```

```python
def unit_standardization(col1, col2, units, values):
    for unit, value in zip(units, values):
        mask = house_copy[col2].str.contains(unit, na= False)
        house_copy.loc[mask, col1] = house_copy.loc[mask, col1] * value

units= ['sqyrd', 'sqm', 'marla', 'acre', 'kanal', 'ground', 'bigha', 'cent']    
carpet_values= [9, 10.764, 272.25, 43560, 5445, 2400, 6806.25, 435.6]
super_values= [9, 10.764, 272.25, 5445, 2400, 340.3125, 72, 43560, 107639.104, 435.6]

unit_standardization('carpet_area', 'carpet_unit', house_copy['carpet_unit'].value_counts().index[1:], carpet_values)
unit_standardization('super_area', 'super_unit', house_copy['super_unit'].value_counts().index[1:], super_values)

```

```python
mask = (house_copy['amount'].isna() == False) & (house_copy['price'].isna() == False) & (house_copy['super_area'].isna())
house_copy.loc[mask, 'super_area'] = (house_copy.loc[mask, 'amount'] * 100000) / house_copy.loc[mask, 'price']
```

```python
mask = house_copy['overlooking'].isna() == False

def strip_advance(l):
    if l is np.nan:
        return np.nan
    for idx, i in enumerate(l):
        l[idx] = i.strip()
    return ",".join(sorted(l)).replace(r',Not Available', '')

house_copy['overlooking'] = house_copy.loc[mask, 'overlooking'].str.split(',').apply(strip_advance)
```

```python
house_copy['overlooking'] = house_copy['overlooking'].astype('category')
```
